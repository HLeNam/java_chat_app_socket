## Thuật toán hồi quy tuyến tính OLS

Mô hình hồi quy tuyến tính có công thức:

$$
\hat{y} = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \dots + \beta_n x_n
$$

hay viết dưới dạng ma trận:

$$
\hat{y} = X \beta
$$

Với:
- $X$ là ma trận kích thước $m \times (n+1)$, trong đó $m$ là số mẫu, $n$ là số đặc trưng, và cột đầu tiên là các giá trị 1 để tính hệ số chệch $\beta_0$
- $\beta$ là vector hệ số cần tìm, kích thước $(n+1) \times 1$
- $\hat{y}$ là vector dự đoán

Mục tiêu là tìm $\beta$ sao cho sai số bình phương giữa $y$ và $\hat{y}$ là nhỏ nhất. Công thức nghiệm đóng của $\beta$ là:

$$
\beta = (X^T X)^{-1} X^T y
$$

---

## Phương pháp đánh giá

### Sai số bình phương trung bình (MSE):

$$
\text{MSE} = \frac{1}{m} \sum_{i=1}^{m} (\hat{y}^{(i)} - y^{(i)})^2
$$

### Hệ số xác định $R^2$:

$$
R^2 = 1 - \frac{SS_{\text{res}}}{SS_{\text{tot}}}
$$

Trong đó:
- $SS_{\text{res}} = \sum_{i=1}^{m} (y^{(i)} - \hat{y}^{(i)})^2$
- $SS_{\text{tot}} = \sum_{i=1}^{m} (y^{(i)} - \bar{y})^2$

---

## Thuật toán Gauss-Jordan để tính ma trận nghịch đảo

Để tính $(X^T X)^{-1}$, ta sử dụng **phương pháp Gauss-Jordan**:

1. Tạo ma trận mở rộng $[A|I]$ (với $A = X^T X$)
2. Dùng phép biến đổi sơ cấp hàng để đưa $A$ về dạng đơn vị
3. Phần bên phải sẽ là ma trận nghịch đảo $A^{-1}$

---

## Cross Validation (cho phần b)

Để chọn 1 đặc trưng tốt nhất:
1. Với mỗi đặc trưng $x_i$, huấn luyện mô hình hồi quy đơn biến $y = \beta_0 + \beta_1 x_i$
2. Tính MSE thông qua k-fold Cross Validation (chia tập dữ liệu thành k phần, huấn luyện và kiểm tra k lần)
3. Chọn đặc trưng có MSE thấp nhất

---

## Kết luận

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
import random
import numpy as np

# ----- Các hàm phụ trợ cho phép tính ma trận -----

def matrix_multiplication(A, B):
    """
    Nhân ma trận A với ma trận B
    """
    if len(A[0]) != len(B):
        raise ValueError("Ma trận không tương thích để nhân")
    
    result = []
    for i in range(len(A)):
        row = []
        for j in range(len(B[0])):
            sum_val = 0
            for k in range(len(B)):
                sum_val += A[i][k] * B[k][j]
            row.append(sum_val)
        result.append(row)
    
    return result

def matrix_transpose(A):
    """
    Chuyển vị ma trận A
    """
    rows = len(A)
    cols = len(A[0])
    
    result = []
    for j in range(cols):
        row = []
        for i in range(rows):
            row.append(A[i][j])
        result.append(row)
    
    return result

def matrix_inverse(A):
    """
    Tính ma trận nghịch đảo bằng phương pháp Gauss-Jordan
    """
    n = len(A)
    
    # Tạo ma trận mở rộng [A|I]
    aug_matrix = []
    for i in range(n):
        row = A[i].copy()
        identity_row = [0] * n
        identity_row[i] = 1
        row.extend(identity_row)
        aug_matrix.append(row)
    
    # Biến đổi thành ma trận bậc thang
    for i in range(n):
        # Tìm phần tử pivot lớn nhất
        max_row = i
        max_val = abs(aug_matrix[i][i])
        
        for j in range(i+1, n):
            if abs(aug_matrix[j][i]) > max_val:
                max_val = abs(aug_matrix[j][i])
                max_row = j
        
        # Đổi hàng
        if max_row != i:
            aug_matrix[i], aug_matrix[max_row] = aug_matrix[max_row], aug_matrix[i]
        
        # Chia hàng pivot để có 1 ở vị trí chéo
        pivot = aug_matrix[i][i]
        if abs(pivot) < 1e-10:
            raise ValueError("Ma trận không khả nghịch (gần singular)")
            
        for j in range(2*n):
            aug_matrix[i][j] /= pivot
        
        # Khử các phần tử khác trong cột
        for j in range(n):
            if j != i:
                factor = aug_matrix[j][i]
                for k in range(2*n):
                    aug_matrix[j][k] -= factor * aug_matrix[i][k]
    
    # Trích xuất ma trận nghịch đảo từ nửa bên phải
    inverse = []
    for i in range(n):
        inverse.append(aug_matrix[i][n:])
    
    return inverse

# ----- Các hàm cho thuật toán hồi quy tuyến tính -----

def add_intercept(X):
    """
    Thêm cột 1 vào đầu ma trận X cho hệ số tự do (intercept)
    """
    result = []
    for i in range(len(X)):
        row = [1]  # Thêm số 1 cho hệ số tự do
        if isinstance(X[i], list):
            row.extend(X[i])
        else:
            row.append(X[i])
        result.append(row)
    
    return result

def compute_coefficient(X, y):
    """
    Tính toán hệ số của mô hình hồi quy tuyến tính sử dụng OLS
    X: ma trận đặc trưng
    y: biến mục tiêu
    Trả về: array của hệ số
    """
    # Thêm cột 1 vào đầu ma trận X (cho intercept)
    X_with_intercept = add_intercept(X)
    
    # Chuyển y từ list thành ma trận cột
    if not isinstance(y[0], list):
        y_matrix = [[val] for val in y]
    else:
        y_matrix = y
    
    # Công thức tính hệ số: β = (X^T * X)^(-1) * X^T * y
    X_transpose = matrix_transpose(X_with_intercept)
    XTX = matrix_multiplication(X_transpose, X_with_intercept)
    XTX_inv = matrix_inverse(XTX)
    XTy = matrix_multiplication(X_transpose, y_matrix)
    
    # Tính beta
    beta = matrix_multiplication(XTX_inv, XTy)
    
    # Chuyển từ ma trận cột thành list 1 chiều
    return [row[0] for row in beta]

def predict(X, coefficients):
    """
    Dự đoán giá trị dựa trên mô hình
    X: ma trận đặc trưng (không có cột intercept)
    coefficients: hệ số của mô hình (bao gồm cả intercept)
    Trả về: array các giá trị dự đoán
    """
    # Thêm cột 1 vào đầu ma trận X (cho intercept)
    X_with_intercept = add_intercept(X)
    
    # Tính y_pred = X * beta
    y_pred = []
    for i in range(len(X_with_intercept)):
        pred = 0
        for j in range(len(coefficients)):
            pred += X_with_intercept[i][j] * coefficients[j]
        y_pred.append(pred)
    
    return y_pred

# ----- Các hàm đánh giá mô hình -----

def mean(values):
    """
    Tính giá trị trung bình
    """
    return sum(values) / len(values)

def sum_squared_error(y_true, y_pred):
    """
    Tính tổng bình phương sai số
    """
    sse = 0
    for i in range(len(y_true)):
        sse += (y_true[i] - y_pred[i]) ** 2
    return sse

def evaluate_model(y_true, y_pred):
    """
    Đánh giá mô hình sử dụng MSE và R^2
    """
    # Tính MSE
    mse = sum_squared_error(y_true, y_pred) / len(y_true)
    
    # Tính R^2
    y_mean = mean(y_true)
    ss_total = sum((y - y_mean) ** 2 for y in y_true)
    ss_residual = sum_squared_error(y_true, y_pred)
    r2 = 1 - (ss_residual / ss_total)
    
    return mse, r2

# ----- Hàm Cross Validation -----

def split_data_for_cv(X, y, n_splits=5):
    """
    Chia dữ liệu thành n_splits phần cho cross validation
    """
    # Shuffle dữ liệu
    combined = list(zip(X, y))
    random.seed(42)  # Đặt seed để kết quả có thể lặp lại
    random.shuffle(combined)
    X_shuffled, y_shuffled = zip(*combined)
    X_shuffled, y_shuffled = list(X_shuffled), list(y_shuffled)
    
    # Chia thành các fold
    fold_size = len(X) // n_splits
    folds = []
    
    for i in range(n_splits):
        start_idx = i * fold_size
        end_idx = start_idx + fold_size if i < n_splits - 1 else len(X)
        test_indices = list(range(start_idx, end_idx))
        train_indices = [j for j in range(len(X)) if j not in test_indices]
        
        X_train = [X_shuffled[j] for j in train_indices]
        y_train = [y_shuffled[j] for j in train_indices]
        X_test = [X_shuffled[j] for j in test_indices]
        y_test = [y_shuffled[j] for j in test_indices]
        
        folds.append((X_train, y_train, X_test, y_test))
    
    return folds

def cross_validation(X, y, feature_idx=None, n_splits=5):
    """
    Thực hiện cross validation
    feature_idx: Index của đặc trưng cần đánh giá, nếu None thì dùng toàn bộ đặc trưng
    """
    folds = split_data_for_cv(X, y, n_splits)
    r2_scores = []
    mse_scores = []
    
    for X_train, y_train, X_test, y_test in folds:
        if feature_idx is not None:
            # Chỉ lấy một đặc trưng
            if isinstance(feature_idx, list):
                # Nhiều đặc trưng
                X_train_selected = [[row[i] for i in feature_idx] for row in X_train]
                X_test_selected = [[row[i] for i in feature_idx] for row in X_test]
            else:
                # Một đặc trưng
                X_train_selected = [row[feature_idx] for row in X_train]
                X_test_selected = [row[feature_idx] for row in X_test]
        else:
            # Dùng tất cả đặc trưng
            X_train_selected = X_train
            X_test_selected = X_test
        
        # Huấn luyện mô hình
        coeffs = compute_coefficient(X_train_selected, y_train)
        
        # Dự đoán và đánh giá
        y_test_pred = predict(X_test_selected, coeffs)
        mse, r2 = evaluate_model(y_test, y_test_pred)
        
        mse_scores.append(mse)
        r2_scores.append(r2)
    
    # Tính trung bình MSE và R^2
    avg_mse = mean(mse_scores)
    avg_r2 = mean(r2_scores)
    
    return avg_mse, avg_r2

# ----- Hàm thêm các đặc trưng đa thức -----

def add_polynomial_features(X, degree=2):
    """
    Thêm các đặc trưng lũy thừa đến bậc degree
    X: ma trận đặc trưng gốc
    degree: bậc lũy thừa cao nhất
    """
    if not isinstance(X[0], list):  # Nếu X chỉ có 1 đặc trưng
        X_poly = []
        for x in X:
            new_row = [x]  # Giữ lại đặc trưng gốc
            for d in range(2, degree+1):
                new_row.append(x ** d)  # Thêm lũy thừa bậc d
            X_poly.append(new_row)
        return X_poly
    else:  # Nếu X có nhiều đặc trưng
        X_poly = []
        for row in X:
            new_row = row.copy()  # Giữ lại các đặc trưng gốc
            for i, val in enumerate(row):
                for d in range(2, degree+1):
                    new_row.append(val ** d)  # Thêm lũy thừa bậc d
            X_poly.append(new_row)
        return X_poly

def add_interaction_features(X):
    """
    Thêm các đặc trưng tương tác (tích các cặp đặc trưng)
    """
    if not isinstance(X[0], list) or len(X[0]) < 2:
        return X  # Không thể thêm tương tác nếu chỉ có 1 đặc trưng
    
    X_interact = []
    for row in X:
        new_row = row.copy()  # Giữ lại các đặc trưng gốc
        
        # Thêm tương tác giữa các cặp đặc trưng
        for i in range(len(row)):
            for j in range(i+1, len(row)):
                new_row.append(row[i] * row[j])
        
        X_interact.append(new_row)
    
    return X_interact

# ----- Phân tích dữ liệu -----

def compute_correlation(feature1, feature2):
    """Tính hệ số tương quan giữa hai đặc trưng"""
    mean1 = mean(feature1)
    mean2 = mean(feature2)
    
    numerator = sum((feature1[i] - mean1) * (feature2[i] - mean2) for i in range(len(feature1)))
    denominator1 = sum((x - mean1) ** 2 for x in feature1) ** 0.5
    denominator2 = sum((x - mean2) ** 2 for x in feature2) ** 0.5
    
    if denominator1 == 0 or denominator2 == 0:
        return 0
    
    return numerator / (denominator1 * denominator2)

def main():
    print("=================================================================")
    print("ĐỒ ÁN TOÁN ỨNG DỤNG VÀ THỐNG KÊ - HỒI QUY TUYẾN TÍNH")
    print("Sinh viên: Hoàng Lê Nam - MSSV: 22120217")
    print("=================================================================\n")
    
    # Đọc dữ liệu
    print("Đang đọc dữ liệu từ file wine.csv...")
    try:
        data = pd.read_csv('wine.csv')
        print("Đọc dữ liệu thành công!")
    except Exception as e:
        print(f"Lỗi khi đọc dữ liệu: {e}")
        return
    
    # Làm sạch tên cột (loại bỏ các dấu ngoặc kép thừa)
    data.columns = [col.replace('"', '') for col in data.columns]
    
    print("\nThông tin dữ liệu:")
    print(f"Kích thước dữ liệu: {data.shape[0]} dòng x {data.shape[1]} cột")
    print(f"Các cột trong dữ liệu: {', '.join(data.columns.tolist())}")
    print("\nXem 5 dòng đầu tiên:")
    print(data.head())
    
    # Tách dữ liệu thành features và target
    X = data.drop('quality', axis=1)
    y = data['quality']
    
    # Chuyển DataFrame thành list
    X_array = X.values.tolist()
    y_array = y.values.tolist()
    
    # ===== YÊU CẦU A: Mô hình với tất cả 11 đặc trưng =====
    print("\n\n=================================================================")
    print("YÊU CẦU A: MÔ HÌNH VỚI TẤT CẢ 11 ĐẶC TRƯNG")
    print("=================================================================")
    
    # Chia dữ liệu thành tập train và test
    print("Đang chia dữ liệu thành tập huấn luyện (80%) và tập kiểm tra (20%)...")
    X_train, X_test, y_train, y_test = train_test_split(X_array, y_array, test_size=0.2, random_state=42)
    
    # Tính toán hệ số từ tập huấn luyện
    print("Đang huấn luyện mô hình với 11 đặc trưng...")
    coefficients = compute_coefficient(X_train, y_train)
    
    # Dự đoán trên tập kiểm tra
    print("Đang thực hiện dự đoán trên tập kiểm tra...")
    y_pred = predict(X_test, coefficients)
    
    # Đánh giá mô hình
    mse, r2 = evaluate_model(y_test, y_pred)
    print(f"\nKết quả mô hình với 11 đặc trưng:")
    print(f"MSE: {mse:.4f}")
    print(f"R^2: {r2:.4f}")
    
    # In ra hệ số của các đặc trưng
    print("\nHệ số của các đặc trưng:")
    feature_names = ['Intercept'] + list(X.columns)
    for name, coef in zip(feature_names, coefficients):
        print(f"{name}: {coef:.4f}")
    
    # Phương trình hồi quy
    equation = f"quality = {coefficients[0]:.4f}"
    for i, name in enumerate(feature_names[1:]):
        if coefficients[i+1] >= 0:
            equation += f" + {coefficients[i+1]:.4f} * {name}"
        else:
            equation += f" - {abs(coefficients[i+1]):.4f} * {name}"
    print("\nPhương trình hồi quy tuyến tính:")
    print(equation)
    
    # ===== YÊU CẦU B: Tìm đặc trưng duy nhất tốt nhất =====
    print("\n\n=================================================================")
    print("YÊU CẦU B: TÌM ĐẶC TRƯNG DUY NHẤT TỐT NHẤT")
    print("=================================================================")
    
    # Đánh giá từng đặc trưng bằng Cross Validation
    feature_names = list(X.columns)
    cv_results = {}
    
    print("Đang đánh giá từng đặc trưng bằng 5-fold Cross Validation...")
    for i, feature in enumerate(feature_names):
        print(f"Đang đánh giá đặc trưng {i+1}/{len(feature_names)}: {feature}")
        avg_mse, avg_r2 = cross_validation(X_array, y_array, feature_idx=i, n_splits=5)
        cv_results[feature] = (avg_mse, avg_r2)
        print(f"  - {feature}: MSE = {avg_mse:.4f}, R^2 = {avg_r2:.4f}")
    
    # Tìm đặc trưng tốt nhất (R^2 cao nhất)
    best_feature = max(cv_results.items(), key=lambda x: x[1][1])
    print(f"\nĐặc trưng tốt nhất: {best_feature[0]}")
    print(f"  - MSE: {best_feature[1][0]:.4f}")
    print(f"  - R^2: {best_feature[1][1]:.4f}")
    
    # Xây dựng mô hình với đặc trưng tốt nhất
    best_idx = feature_names.index(best_feature[0])
    X_best = [row[best_idx] for row in X_array]
    
    # Chia tập train/test
    X_train_best, X_test_best, y_train_best, y_test_best = train_test_split(
        X_best, y_array, test_size=0.2, random_state=42
    )
    
    # Tính toán hệ số
    best_coeffs = compute_coefficient(X_train_best, y_train_best)
    
    # Dự đoán và đánh giá
    y_pred_best = predict(X_test_best, best_coeffs)
    mse_best, r2_best = evaluate_model(y_test_best, y_pred_best)
    
    print(f"\nMô hình với đặc trưng {best_feature[0]}:")
    print(f"MSE: {mse_best:.4f}")
    print(f"R^2: {r2_best:.4f}")
    print(f"Phương trình: quality = {best_coeffs[0]:.4f} + {best_coeffs[1]:.4f} * {best_feature[0]}")
    
    # ===== YÊU CẦU C: Xây dựng mô hình tối ưu =====
    print("\n\n=================================================================")
    print("YÊU CẦU C: XÂY DỰNG MÔ HÌNH TỐI ƯU")
    print("=================================================================")
    
    # Chọn top N đặc trưng tốt nhất
    sorted_features = sorted(cv_results.items(), key=lambda x: x[1][1], reverse=True)
    
    print("1. Tìm số lượng đặc trưng tối ưu...")
    
    # Thử các số lượng đặc trưng khác nhau
    best_num_features = 0
    best_r2 = -float('inf')
    best_mse = float('inf')
    best_feature_set = []
    
    for n in range(1, len(sorted_features) + 1):
        selected_features = [feature for feature, _ in sorted_features[:n]]
        selected_indices = [feature_names.index(feature) for feature in selected_features]
        
        print(f"Thử nghiệm với {n} đặc trưng: {', '.join(selected_features)}")
        
        # Đánh giá bằng cross validation
        avg_mse, avg_r2 = cross_validation(X_array, y_array, feature_idx=selected_indices, n_splits=5)
        print(f"  - MSE: {avg_mse:.4f}, R^2: {avg_r2:.4f}")
        
        # Cập nhật nếu tìm thấy mô hình tốt hơn
        if avg_r2 > best_r2:
            best_r2 = avg_r2
            best_mse = avg_mse
            best_num_features = n
            best_feature_set = selected_features
    
    print(f"\nSố lượng đặc trưng tối ưu: {best_num_features}")
    print(f"Tập đặc trưng tối ưu: {', '.join(best_feature_set)}")
    print(f"MSE: {best_mse:.4f}")
    print(f"R^2: {best_r2:.4f}")
    
    # Xây dựng mô hình tuyến tính với tập đặc trưng tối ưu
    best_indices = [feature_names.index(feature) for feature in best_feature_set]
    X_selected = [[row[i] for i in best_indices] for row in X_array]
    
    # Chia tập train/test
    X_train_selected, X_test_selected, y_train_selected, y_test_selected = train_test_split(
        X_selected, y_array, test_size=0.2, random_state=42
    )
    
    # Tính toán hệ số
    selected_coeffs = compute_coefficient(X_train_selected, y_train_selected)
    
    # Dự đoán và đánh giá
    y_pred_selected = predict(X_test_selected, selected_coeffs)
    mse_selected, r2_selected = evaluate_model(y_test_selected, y_pred_selected)
    
    print(f"\n2. Mô hình tuyến tính tối ưu với {best_num_features} đặc trưng:")
    print(f"MSE: {mse_selected:.4f}")
    print(f"R^2: {r2_selected:.4f}")
    
    # In ra hệ số
    print("\nPhương trình hồi quy:")
    feature_names_selected = ['Intercept'] + best_feature_set
    equation = f"quality = {selected_coeffs[0]:.4f}"
    for i, (name, coef) in enumerate(zip(feature_names_selected[1:], selected_coeffs[1:])):
        if coef >= 0:
            equation += f" + {coef:.4f} * {name}"
        else:
            equation += f" - {abs(coef):.4f} * {name}"
    print(equation)
    
    # Thử nghiệm với mô hình đa thức cho đặc trưng tốt nhất
    print("\n3. Thử nghiệm mô hình đa thức với đặc trưng tốt nhất...")
    
    # Tạo các đặc trưng đa thức bậc 2 cho đặc trưng tốt nhất
    X_poly = add_polynomial_features(X_best, degree=2)
    
    # Chia tập train/test
    X_train_poly, X_test_poly, y_train_poly, y_test_poly = train_test_split(
        X_poly, y_array, test_size=0.2, random_state=42
    )
    
    # Tính toán hệ số
    poly_coeffs = compute_coefficient(X_train_poly, y_train_poly)
    
    # Dự đoán và đánh giá
    y_pred_poly = predict(X_test_poly, poly_coeffs)
    mse_poly, r2_poly = evaluate_model(y_test_poly, y_pred_poly)
    
    print(f"Mô hình đa thức bậc 2 với đặc trưng {best_feature[0]}:")
    print(f"MSE: {mse_poly:.4f}")
    print(f"R^2: {r2_poly:.4f}")
    print(f"Phương trình: quality = {poly_coeffs[0]:.4f} + {poly_coeffs[1]:.4f} * {best_feature[0]} + {poly_coeffs[2]:.4f} * {best_feature[0]}^2")
    
    # Thử nghiệm mô hình với tương tác giữa các đặc trưng tốt nhất
    if len(best_feature_set) >= 2:
        print("\n4. Thử nghiệm mô hình với tương tác giữa các đặc trưng tốt nhất...")
        
        # Tạo mô hình với tương tác
        X_interact = add_interaction_features(X_selected)
        
        # Chia tập train/test
        X_train_interact, X_test_interact, y_train_interact, y_test_interact = train_test_split(
            X_interact, y_array, test_size=0.2, random_state=42
        )
        
        # Tính toán hệ số
        interact_coeffs = compute_coefficient(X_train_interact, y_train_interact)
        
        # Dự đoán và đánh giá
        y_pred_interact = predict(X_test_interact, interact_coeffs)
        mse_interact, r2_interact = evaluate_model(y_test_interact, y_pred_interact)
        
        print(f"Mô hình với tương tác giữa các đặc trưng tốt nhất:")
        print(f"MSE: {mse_interact:.4f}")
        print(f"R^2: {r2_interact:.4f}")
    
    # ===== Trực quan hóa kết quả =====
    print("\n\n=================================================================")
    print("TRỰC QUAN HÓA KẾT QUẢ")
    print("=================================================================")
    
    # 1. Biểu đồ cột so sánh R^2 của các mô hình
    models = ['11 đặc trưng', f'1 đặc trưng\n({best_feature[0]})', f'{best_num_features} đặc trưng\n(tối ưu)']
    r2_values = [r2, r2_best, r2_selected]
    
    if 'r2_poly' in locals():
        models.append(f'Đa thức bậc 2\n({best_feature[0]})')
        r2_values.append(r2_poly)
    
    if 'r2_interact' in locals():
        models.append('Mô hình\nvới tương tác')
        r2_values.append(r2_interact)
    
    plt.figure(figsize=(12, 6))
    bars = plt.bar(models, r2_values, color=['blue', 'green', 'orange', 'red', 'purple'][:len(models)])
    plt.ylim(0, max(r2_values) * 1.2)
    plt.ylabel('Hệ số xác định R²', fontsize=12)
    plt.title('So sánh hệ số xác định (R²) của các mô hình', fontsize=14)
    
    # Thêm nhãn giá trị trên các cột
    for bar, val in zip(bars, r2_values):
        plt.text(bar.get_x() + bar.get_width()/2, val + 0.01, f'{val:.4f}', 
                 ha='center', va='bottom', fontweight='bold')
    
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.savefig('model_r2_comparison.png')
    print("Đã lưu biểu đồ so sánh R² vào file 'model_r2_comparison.png'")
    plt.show()
    
    # 2. Biểu đồ cột so sánh MSE của các mô hình
    mse_values = [mse, mse_best, mse_selected]
    
    if 'mse_poly' in locals():
        mse_values.append(mse_poly)
    
    if 'mse_interact' in locals():
        mse_values.append(mse_interact)
    
    plt.figure(figsize=(12, 6))
    bars = plt.bar(models, mse_values, color=['blue', 'green', 'orange', 'red', 'purple'][:len(models)])
    plt.ylim(0, max(mse_values) * 1.2)
    plt.ylabel('Mean Squared Error (MSE)', fontsize=12)
    plt.title('So sánh sai số bình phương trung bình (MSE) của các mô hình', fontsize=14)
    
    # Thêm nhãn giá trị trên các cột
    for bar, val in zip(bars, mse_values):
        plt.text(bar.get_x() + bar.get_width()/2, val + 0.01, f'{val:.4f}', 
                 ha='center', va='bottom', fontweight='bold')
    
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.savefig('model_mse_comparison.png')
    print("Đã lưu biểu đồ so sánh MSE vào file 'model_mse_comparison.png'")
    plt.show()
    
    # 3. Trực quan hóa mô hình đơn biến tốt nhất
    plt.figure(figsize=(10, 6))
    plt.scatter(X_test_best, y_test_best, alpha=0.5, label='Dữ liệu thực tế')
    
    # Sắp xếp các điểm để vẽ đường hồi quy
    points = sorted(zip(X_test_best, y_pred_best))
    x_sorted, y_sorted = zip(*points)
    
    plt.plot(x_sorted, y_sorted, color='red', linewidth=2, label='Đường hồi quy tuyến tính')
    
    # Nếu có mô hình đa thức, vẽ thêm đường hồi quy đa thức
    if 'y_pred_poly' in locals():
        points_poly = sorted(zip(X_test_poly, y_pred_poly), key=lambda x: x[0][0])
        x_poly = [p[0][0] for p in points_poly]  # Lấy giá trị đặc trưng gốc
        y_poly = [p[1] for p in points_poly]
        plt.plot(x_poly, y_poly, color='green', linewidth=2, linestyle='--', label='Đường hồi quy đa thức bậc 2')
    
    plt.xlabel(best_feature[0], fontsize=12)
    plt.ylabel('Chất lượng rượu vang', fontsize=12)
    plt.title(f'Mô hình hồi quy với đặc trưng {best_feature[0]}', fontsize=14)
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.savefig('best_feature_model.png')
    print(f"Đã lưu biểu đồ mô hình với đặc trưng {best_feature[0]} vào file 'best_feature_model.png'")
    plt.show()
    
    # 4. Trực quan hóa dự đoán vs giá trị thực tế cho mô hình 11 đặc trưng
    plt.figure(figsize=(10, 6))
    plt.scatter(y_test, y_pred, alpha=0.5)
    plt.plot([min(y_test), max(y_test)], [min(y_test), max(y_test)], 'r--', linewidth=2)
    plt.xlabel('Giá trị thực tế', fontsize=12)
    plt.ylabel('Giá trị dự đoán', fontsize=12)
    plt.title('Mô hình với tất cả 11 đặc trưng: Dự đoán vs Thực tế', fontsize=14)
    plt.grid(True)
    plt.tight_layout()
    plt.savefig('full_model_prediction.png')
    print("Đã lưu biểu đồ dự đoán vs thực tế cho mô hình đầy đủ vào file 'full_model_prediction.png'")
    plt.show()
    
    # 5. Trực quan hóa dự đoán vs giá trị thực tế cho mô hình tối ưu
    plt.figure(figsize=(10, 6))
    plt.scatter(y_test_selected, y_pred_selected, alpha=0.5)
    plt.plot([min(y_test_selected), max(y_test_selected)], [min(y_test_selected), max(y_test_selected)], 'r--', linewidth=2)
    plt.xlabel('Giá trị thực tế', fontsize=12)
    plt.ylabel('Giá trị dự đoán', fontsize=12)
    plt.title(f'Mô hình tối ưu ({best_num_features} đặc trưng): Dự đoán vs Thực tế', fontsize=14)
    plt.grid(True)
    plt.tight_layout()
    plt.savefig('optimal_model_prediction.png')
    print(f"Đã lưu biểu đồ dự đoán vs thực tế cho mô hình tối ưu vào file 'optimal_model_prediction.png'")
    plt.show()
    
    # 6. Biểu đồ hộp (Boxplot) cho các đặc trưng quan trọng nhất
    top_features = best_feature_set[:min(3, len(best_feature_set))]
    top_indices = [feature_names.index(feature) for feature in top_features]
    
    plt.figure(figsize=(12, 6))
    box_data = [
        [row[idx] for row in X_array] 
        for idx in top_indices
    ]
    plt.boxplot(box_data, labels=top_features)
    plt.title('Phân phối của các đặc trưng quan trọng nhất', fontsize=14)
    plt.ylabel('Giá trị', fontsize=12)
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.savefig('top_features_boxplot.png')
    print("Đã lưu biểu đồ phân phối đặc trưng vào file 'top_features_boxplot.png'")
    plt.show()
    
    # 7. Phân tích độ quan trọng của các đặc trưng
    print("\nPhân tích độ quan trọng của các đặc trưng dựa trên hệ số tương quan với chất lượng:")
    
    # Tính hệ số tương quan với chất lượng
    feature_importance = []
    for i, feature in enumerate(feature_names):
        corr = compute_correlation([row[i] for row in X_array], y_array)
        feature_importance.append((feature, abs(corr), corr))
    
    # Sắp xếp theo độ quan trọng (giá trị tuyệt đối của hệ số tương quan)
    feature_importance.sort(key=lambda x: x[1], reverse=True)
    
    # In độ quan trọng
    for name, abs_corr, corr in feature_importance:
        correlation_type = "dương" if corr > 0 else "âm"
        print(f"{name}: |r| = {abs_corr:.4f}, r = {corr:.4f} (tương quan {correlation_type})")
    
    # Vẽ biểu đồ độ quan trọng
    plt.figure(figsize=(12, 6))
    features = [f[0] for f in feature_importance]
    importances = [f[1] for f in feature_importance]
    colors = ['green' if f[2] > 0 else 'red' for f in feature_importance]
    
    bars = plt.barh(range(len(features)), importances, align='center', color=colors)
    plt.yticks(range(len(features)), features)
    plt.xlabel('Độ quan trọng (Hệ số tương quan tuyệt đối)', fontsize=12)
    plt.title('Độ quan trọng của các đặc trưng đối với chất lượng rượu vang', fontsize=14)
    
    # Thêm nhãn giá trị và loại tương quan
    for i, bar in enumerate(bars):
        corr_value = feature_importance[i][2]
        corr_type = "(+)" if corr_value > 0 else "(-)"
        plt.text(bar.get_width() + 0.01, bar.get_y() + bar.get_height()/2, 
                 f'{importances[i]:.4f} {corr_type}', va='center')
    
    plt.grid(axis='x', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.savefig('feature_importance.png')
    print("Đã lưu biểu đồ độ quan trọng đặc trưng vào file 'feature_importance.png'")
    plt.show()
    
    # ===== Kết luận =====
    print("\n\n=================================================================")
    print("KẾT LUẬN")
    print("=================================================================")
    
    print("\n1. So sánh các mô hình:")
    model_comparison = [
        ("Mô hình với 11 đặc trưng", mse, r2),
        (f"Mô hình với 1 đặc trưng ({best_feature[0]})", mse_best, r2_best),
        (f"Mô hình tối ưu ({best_num_features} đặc trưng)", mse_selected, r2_selected)
    ]
    
    if 'mse_poly' in locals():
        model_comparison.append((f"Mô hình đa thức bậc 2 với {best_feature[0]}", mse_poly, r2_poly))
    
    if 'mse_interact' in locals():
        model_comparison.append(("Mô hình với tương tác", mse_interact, r2_interact))
    
    # Sắp xếp theo R^2 giảm dần
    model_comparison.sort(key=lambda x: x[2], reverse=True)
    
    print("\nXếp hạng mô hình theo R^2:")
    for i, (model_name, model_mse, model_r2) in enumerate(model_comparison):
        print(f"{i+1}. {model_name}: R^2 = {model_r2:.4f}, MSE = {model_mse:.4f}")
    
    # Mô hình tốt nhất
    best_model = model_comparison[0]
    print(f"\nMô hình tốt nhất: {best_model[0]} với R^2 = {best_model[2]:.4f}, MSE = {best_model[1]:.4f}")
    
    print("\n2. Đặc trưng quan trọng nhất:")
    for i, (feature, abs_corr, corr) in enumerate(feature_importance[:3]):
        correlation_type = "dương" if corr > 0 else "âm"
        print(f"{i+1}. {feature}: |r| = {abs_corr:.4f}, r = {corr:.4f} (tương quan {correlation_type})")
    
    print("\n3. Nhận xét:")
    print(f"- Mô hình tối ưu sử dụng {best_num_features} đặc trưng đạt hiệu suất tốt nhất")
    print(f"- Đặc trưng {feature_importance[0][0]} có ảnh hưởng lớn nhất đến chất lượng rượu vang")
    print("- Hiệu suất của các mô hình còn khá thấp (R^2 < 0.5), cho thấy cần thêm dữ liệu hoặc đặc trưng khác")
    print("- Các đặc trưng có tương quan cao nhất với chất lượng rượu vang là:", end=" ")
    print(", ".join([f[0] for f in feature_importance[:3]]))
    
    print("\n=================================================================")
    print("ĐỒ ÁN HOÀN THÀNH!")
    print("=================================================================")

if __name__ == "__main__":
    main()
```
